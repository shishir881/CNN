```python
def add(x, y):
    return x + y
def sub(x, y):
    return x - y
def mult(x, y):     
    return x * y
def div(x, y):
    return x / y
def modulu(x, y):
    return x % y    
a=int(input("Enter a number: "))
b=int(input("Enter another number: "))
calc=input("Enter the operation you want to perform(+,-,*,/,%): ")
print("The numbers you entered are:", a, "and", b)
match calc:
    case '+':
        print("The addition of two numbers is:", add(a,b))
    case '-':
        print("The subtraction of two numbers is:", sub(a,b))
    case '*':
        print("The multiplication of two numbers is:", mult(a,b))
    case '/':
        print("The division of two numbers is:", div(a,b))
    case '%':
        print("The modulus of two numbers is:", modulu(a,b))
    case _:
        print("Invalid operation")

```

```python

def add_numbers(num_list):
    total = 0
    for n in num_list:
        total += n
    return total


def multiply_numbers(num_list):
    product = 1
    for n in num_list:
        product *= n
    return product


def main():
    # Take tuple input from user
    user_input = input("Enter numbers separated by commas: ")   # e.g., 2,3,4

    # Convert input to tuple of integers
    num_tuple = tuple(map(int, user_input.split(',')))

    print("Tuple entered by user:", num_tuple)

    # Convert tuple into list
    num_list = list(num_tuple)
    print("Converted list:", num_list)

    # Perform addition and multiplication
    addition = add_numbers(num_list)
    multiplication = multiply_numbers(num_list)

    # Display results
    print("Addition of numbers:", addition)
    print("Multiplication of numbers:", multiplication)


# Run the program
main()
```

```python
file_name="sample.txt"
f=open(file_name, 'r')
content=f.read()
freq={}
for char in content:
    if char in freq:
        freq[char]+=1
    else:
        freq[char]=1
print("Character Frequency in file:")
for key, value in freq.items():
    print(f"'{key}': {value}")
f.close()
```

```python
# read each line of a file in reverse order
file_name="sample.txt"
f=open(file_name, 'r')
lines=f.readlines()
for line in lines:
    line = line.strip()      # remove newline character
    reversed_line = line[::-1]  # reverse the line
    print(reversed_line)
f.close()
```

```python
file_name="sample.txt"
f=open(file_name, 'r')
lines=f.readlines()
n_lines=len(lines)
n_words=0
n_chars=0
for line in lines:
    line = line.strip()
    for char in line:
        if char != ' ':
            n_chars+=1
    words=line.split()
    n_words+=len(words)
print("Number of lines:", n_lines)
print("Number of words:", n_words)
print("Number of characters (excluding spaces):", n_chars)
f.close()
```

```python
file_name="poems.txt"
f=open(file_name, 'r')
content=f.read()
n_twinkle=0
n_star=0
for word in content.split():
    if(word.endswith(",") or word.endswith(".") or word.endswith("!") or word.endswith("?")):
        word=word[:-1]  # Remove punctuation from the end of the word
    if(word.upper()=="TWINKLE"):
        n_twinkle+=1
    elif(word.upper()=="STAR"):
        n_star+=1
print("Number of times 'TWINKLE' appears:", n_twinkle)
print("Number of times 'STAR' appears:", n_star)
f.close()
```
